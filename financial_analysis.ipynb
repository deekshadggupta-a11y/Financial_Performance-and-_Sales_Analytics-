```python
#convert date column
df['Order Date']=pd.to_datetime(df['Order Date'], format='mixed')
df['Ship Date']=pd.to_datetime(df['Ship Date'], format='mixed')
```

```python
df.dtypes
```

```python
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month
df['Order Quarter'] = df['Order Date'].dt.quarter
```

```python
df['Profit Margin']= (df['Profit'] / df['Sales']*100).round(2)
df['Delivery Days']= (df['Ship Date'] - df['Order Date']).dt.days
```

```python
print(df[['Profit Margin', 'Delivery Days','Order Year','Order Month','Order Quarter']].head())
```

```python
#sales by category
category_sales=df.groupby('Category').agg(
    TotalSales=('Sales', 'sum'),
    TotalProfit=('Profit', 'sum'),
    AvgProfitMargin=('Profit Margin', 'mean'),
    OrderCount=('Order ID', 'count')
).round(2).reset_index()

print('Sales By Category:')
print(category_sales)   
```

```python
#profit by margin
Region_profit=df.groupby('Region').agg(
    TotalSales=('Sales', 'sum'),
    TotalProfit=('Profit', 'sum'),
    AvgProfitMargin=('Profit Margin', 'mean'),
    AvgDiscount=('Discount', 'mean')
).round(2).reset_index()

print('Profit By Region:')
print(Region_profit)    
```

```python
#correlation between discount and profit
correlation=df['Discount'].corr(df['Profit Margin'])
print(f"Correlation between Discount and Profit Margin: {correlation:.3f}")

#correlation matrix
corr_matrix = df[['Sales', 'Profit', 'Discount', 'Quantity', 'Profit Margin']].corr().round(3)
print("Correlation Matrix:")
print(corr_matrix)
```

```python
#monthly sales trend
monthly_sales = df.groupby(['Order Year','Order Month']).agg(
    TotalSales=('Sales', 'sum'),
    TotalProfit=('Profit', 'sum')
).round(2).reset_index()

monthly_sales['YearMonth']=monthly_sales['Order Year'].astype(str) + '-' + monthly_sales['Order Month'].astype(str).str.zfill(2)

print('Monthly Sales Trend:')
print(monthly_sales.head())
```

```python
#discount analysis by region
discount_region=df.groupby('Region').agg(
    AvgDiscount=('Discount', 'mean'),
    AvgProfitMargin=('Profit Margin', 'mean'),
    LossOrders=('Profit', lambda x: (x < 0).sum()),
    TotalOrders=('Order ID', 'count')
).round(2).reset_index()

discount_region['LossRate%']= (discount_region['LossOrders'] / discount_region['TotalOrders']*100).round(2)

print('Discount Analysis by Region:')
print(discount_region)
```

```python
plt.style.use('seaborn-v0_8-whitegrid')
fig, axes=plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Financial Performance Analysis - Superstore', fontsize=16, fontweight='bold', y=1.02)

#chart1- sales by category
category_sales.plot(kind='bar', x='Category', y='TotalSales', ax=axes[0,0], color='steelblue', legend=False)
axes[0,0].set_title('Total Sales by Category')
axes[0,0].set_xlabel('Category')
axes[0,0].set_ylabel('Total Sales ($)')
axes[0,0].tick_params(axis='x', rotation=0)


#chart2- profit marginby region
colors= ['red' if x<0 else 'steelblue' for x in discount_region['AvgProfitMargin']]
axes[0,1].bar(discount_region['Region'], discount_region['AvgProfitMargin'], color=colors)
axes[0,1].set_title('Avg Profit Margin % by Region')
axes[0,1].set_xlabel('Region')
axes[0,1].set_ylabel('Avg Profit Margin %')
axes[0,1].axhline(y=0, color='black', linestyle='-', linewidth=0.5)


# Chart 3 — Discount vs Profit Margin Scatter
axes[1,0].scatter(df['Discount'], df['Profit Margin'], 
                  alpha=0.3, color='steelblue', s=10)
axes[1,0].set_title(f'Discount vs Profit Margin (r={correlation:.3f})')
axes[1,0].set_xlabel('Discount')
axes[1,0].set_ylabel('Profit Margin %')


# Chart 4 — Loss Rate by Region
colors2 = ['red' if x > 20 else 'steelblue' for x in discount_region['LossRate%']]
axes[1,1].bar(discount_region['Region'], discount_region['LossRate%'], color=colors2)
axes[1,1].set_title('Loss Making Orders % by Region')
axes[1,1].set_xlabel('Region')
axes[1,1].set_ylabel('Loss Rate %')

plt.tight_layout()
plt.savefig('financial_analysis_charts.png', dpi=150, bbox_inches='tight')
plt.show()
print("Charts saved!")
```

```python
# Monthly trend
fig, ax = plt.subplots(figsize=(14, 5))

for year in sorted(df['Order Year'].unique()):
    yearly_data = monthly_sales[monthly_sales['Order Year'] == year]
    ax.plot(yearly_data['Order Month'], 
            yearly_data['TotalSales'],
            marker='o', label=str(year), linewidth=2)

ax.set_title('Monthly Sales Trend by Year', fontsize=14, fontweight='bold')
ax.set_xlabel('Month')
ax.set_ylabel('Total Sales ($)')
ax.set_xticks(range(1, 13))
ax.set_xticklabels(['Jan','Feb','Mar','Apr','May','Jun',
                     'Jul','Aug','Sep','Oct','Nov','Dec'])
ax.legend(title='Year')
plt.tight_layout()
plt.savefig('monthly_trend.png', dpi=150, bbox_inches='tight')
plt.show()
print("Monthly trend saved!")
```
